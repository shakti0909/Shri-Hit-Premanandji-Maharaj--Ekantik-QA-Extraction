```python
from bs4 import XMLParsedAsHTMLWarning
import warnings

warnings.filterwarnings("ignore", category=XMLParsedAsHTMLWarning)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(markup=vtt_decoded, features="html.parser")
```

```python
soup.find_all("text")  # finds all <text> tags

tracnscript_text_block = ""
for tag in soup.find_all("text"):
    start_sv = int(tag['start'].split(".")[0])
    tracnscript_text_block = ""
    if start_sv>139:
        tracnscript_text_block += tag.text
        if start_sv>468:
            break
```

```python
import webvtt
from io import StringIO

vtt = webvtt.read_buffer(StringIO(vtt_decoded))
for caption in vtt:
    # print(int(caption.start.split(":")[2].split(".")[0]))
    if int(caption.start.split(":")[2].split(".")[0])>149 :
        print(caption.start)
    # if caption.start>149.0 and caption.end>448.0:
        print(f"[{caption.start} - {caption.end}] {caption.text}")
        if int(caption.start.split(":")[2].split(".")[0])<448:
             break
```

```python
from bs4 import BeautifulSoup

# First install BeautifulSoup if you haven't:
# pip install beautifulsoup4

# Sample HTML content
html_content  = html
# Create BeautifulSoup object
soup = BeautifulSoup(html_content, 'html.parser')

# Method 1: Find by multiple classes using CSS selector
element = soup.select_one('h4.macro-markers.style-scope')

# Get text and title attribute
if element:
    text = element.text.strip()
    title = element.get('title')
    print(f"Text: {text}")
    print(f"Title: {title}")
```
