# Prototype 2: Playlist and Video Link Extraction

This notebook demonstrates extracting playlists from a YouTube channel based on a keyword, then extracting video links from those playlists.

```python
import sys
import os

# Add scripts directory to path
scripts_dir = os.path.abspath(os.path.join(os.getcwd(), '..', 'scripts'))
sys.path.insert(0, scripts_dir)

# Import functions directly from modules
import importlib.util

# Load playlists_extract module
spec = importlib.util.spec_from_file_location("playlists_extract", os.path.join(scripts_dir, "playlists_extract.py"))
playlists_extract = importlib.util.module_from_spec(spec)
spec.loader.exec_module(playlists_extract)

# Load qa_extract module
spec = importlib.util.spec_from_file_location("qa_extract", os.path.join(scripts_dir, "qa_extract.py"))
qa_extract = importlib.util.module_from_spec(spec)
spec.loader.exec_module(qa_extract)

# Extract functions
get_playlists_by_channel_and_keyword = playlists_extract.get_playlists_by_channel_and_keyword
save_playlists_to_csv = playlists_extract.save_playlists_to_csv
get_playlist_videos = qa_extract.get_playlist_videos

import pandas as pd
```

```python
# Input: Channel name (e.g., '@ChannelName') and keyword
channel_name =  "@BhajanMarg"#input("Enter channel name (e.g., @ChannelName): ")
keyword = "Ekantik"#input("Enter keyword to filter playlists: ")
```

```python
# Extract playlists matching the keyword
playlists = get_playlists_by_channel_and_keyword(channel_name, keyword)

# Save to CSV
save_playlists_to_csv(playlists, "../data/playlists_filtered.csv")
```

```python
# Read the playlists CSV
import time
playlists_df = pd.read_csv("../data/playlists_filtered.csv")

# Import the parallel extraction function
extract_videos_parallel = qa_extract.extract_videos_parallel

# Extract videos in parallel (3 workers = good balance for YouTube rate limits)
all_video_links, failed_playlists = extract_videos_parallel(playlists_df, max_workers=3)

# Save video links to CSV
video_links_df = pd.DataFrame(all_video_links)
video_links_df.to_csv("../data/video_links.csv", index=False)

# Summary
print(f"\n{'='*60}")
print(f"✅ EXTRACTION COMPLETE")
print(f"{'='*60}")
print(f"Total playlists processed: {len(playlists_df)}")
print(f"Successful playlists: {len(playlists_df) - len(failed_playlists)}")
print(f"Failed/partial playlists: {len(failed_playlists)}")
print(f"Total videos extracted: {len(video_links_df)}")
print(f"File saved: ../data/video_links.csv")

if failed_playlists:
    print(f"\n⚠️  Failed playlists:")
    for title, reason in failed_playlists:
        print(f"  - {title}: {reason}")
```
